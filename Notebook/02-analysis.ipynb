```python
import os, re
import ollama
import pandas as pd
from dotenv import load_dotenv
from sqlalchemy import create_engine, inspect, text
from sqlalchemy.engine import URL

load_dotenv("../Notebook/.env", override=True)

for name in ["AGENT_USER", "AGENT_PASSWORD", "DB_NAME"]:
    if not os.getenv(name):
        raise RuntimeError(f"{name} fehlt in der .env")

agent_engine = create_engine(URL.create(
    "postgresql+psycopg2",
    username=os.getenv("AGENT_USER"),
    password=os.getenv("AGENT_PASSWORD"),
    host="localhost",
    port=5433,
    database=os.getenv("DB_NAME"),
))

with agent_engine.connect() as conn:
    print(conn.execute(text("SELECT current_user")).scalar())
```
